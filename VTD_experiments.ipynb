```python
import torch
import torch.optim as optim
import os
import utils.data_load
import numpy as np
from VTD_modules import VTD,trainer

# hyperparameter
input_dim = 30
head=4
hidden_dim = 256
latent_dim = 256
output_dim = 1
treatment_dim = 3
length = 30
batch_size = 64
epochs = 10 #sampled
lr = 1e-4
weight_decay = 1e-5
alpha = 50 #adjust the outcome loss
```

```python
device = torch.device('cuda')

torch.manual_seed(np.random.randint(3407))

print('-------Training-----')
train_loader, val_loader, test_loader = utils.data_load.loader_all('syn_data', 7, batch_size)
#build model

if not os.path.exists('./checkpoints'):
    os.makedirs('./checkpoints')
model_save_path = './checkpoints/best_model_'+str(epochs)+'epoch.pth'
model = VTD(input_dim, hidden_dim, latent_dim, output_dim, treatment_dim, head, length)
model.to(device)
optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)

trainer(model, train_loader, val_loader, test_loader, optimizer, epochs, alpha, model_save_path, resume=0)
```

```python
from utils.vtd_utils import evaluate_syn
dataset = 'syn'
all_predicted_outcomes,all_true_outcomes,all_masks,ite_samples_original=evaluate_syn(model, dataset,test_loader)
```
